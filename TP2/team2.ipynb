## My Shopping List

# A list can grow and shrink. Build a shopping list and change it step by step. Print the list after each step 
to see what happened.

**Step 1**: Create the list with three items and print it

```python
shopping =["rice","eggs","milk"]

print(shopping)
```

**Step 2**: . Add "bread" at the **end** with append(), and "water" at the **front** (position 0) with insert()

```python
shopping.append("bread")
shopping.insert(0,"water")
print("Added :",shopping)
```

**Step 3**: Remove "eggs" by its value with remove(). Then remove the **last** item with pop(), which also 
gives it back to you.

```python
shopping.remove("eggs")
last =shopping.pop()
print("Removed :",last)
print("Now :",shopping)
```

**Step 4**: Use slicing list[start:stop:step] to print the first 2 items, the last 2 items and the reversed 
list.

```python
print("Fist 2 : ",shopping[0:2])
print("last 2: ",shopping[1:3])
print("Reversed : ",shopping[::-1])
```

**Step 5**: Check whether "milk" is in the list with in, then print each item with a number using 
enumerate().

```python
print("Is milk in the last?","milk" in shopping)
for i , items in enumerate(shopping,start=1):
    print(i ,items)
```
